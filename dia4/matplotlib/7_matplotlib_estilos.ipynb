```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,5,11)
```

# ESTILOS DE PYPLOT

```python
plt.style.available
```

```python
plt.style.use('dark_background')
```

```python
fig,axes = plt.subplots(figsize=(8,8))
axes.plot(x,x+1,'r--')
fig.show()
```

```python
fig,axes = plt.subplots(figsize=(8,8))
axes.plot(x,x+1,color='#15B01A',linestyle=':',marker='x',linewidth=5)
axes.plot(x,x+2,color='red',linestyle='--',marker='8',linewidth=9)
axes.plot(x,x+3,color='yellow',linestyle='dashed',marker='x',linewidth=3)
axes.plot(x,x+4,color='pink',linestyle='-',marker='x',linewidth=2)
fig.show()
```
