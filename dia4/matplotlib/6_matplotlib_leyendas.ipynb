```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(0,5,11)
y = np.sin(x)
```

# LEYENDAS CON MATPLOTLIB

## PARA QUE SE MUESTREN LAS LEYENDAS SE DEBE LLAMAR A ESTA FUNCIÓN plt.legend()

```python
plt.plot(x,y,label='sin(x)')
plt.title('Relación de X y Y')
plt.xlabel('VARIABLE X')
plt.ylabel('VARIABLE Y')
plt.legend()
plt.show()
```

# LEYENDAS CON POO

```python
fig,axes = plt.subplots(1,2,figsize=(5,5))
axes[0].plot(x,y,label='SIN(X)')
axes[0].set_title('Relación de X y Y')
axes[0].set_xlabel('VARIABLE X')
axes[0].set_ylabel('VARIABLE Y')
axes[0].legend()
axes[1].plot(y,x,label='YX')
axes[1].set_title('Relación de Y y X')
axes[1].set_xlabel('VARIABLE Y')
axes[1].set_ylabel('VARIABLE X')
axes[1].legend()

plt.show()
```
