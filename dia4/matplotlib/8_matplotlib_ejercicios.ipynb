```python
import pandas as pd
df_books = pd.read_csv('https://github.com/cesarmayta/datasets/raw/refs/heads/main/libros.csv')
df_books.head(2)
```

📌 **Ejercicio 1: Evolución del precio promedio por año**

Muestra cómo ha cambiado el precio promedio de los libros en el tiempo.


```python
import matplotlib.pyplot as plt
```

```python
df_eje1 = df_books.groupby('Year')['Price'].mean().reset_index()
df_eje1
```

```python
x = df_eje1['Year']
y = df_eje1['Price']
plt.plot(x,y,color='green',label='Evolucion Precio')
plt.title('Precio promedio por año')
plt.xlabel('Año')
plt.ylabel('Precio Promedio')
plt.legend()
plt.show()
```

📌 **Ejercicio 2: Comparar calificación promedio por género**
Compara la calificación promedio entre libros de ficción y no ficción.

```python
df_eje2 = df_books.groupby('Genre')['User Rating'].mean().reset_index()
df_eje2
```

```python
x = df_eje2['Genre']
y = df_eje2['User Rating']
plt.bar(x,y,color=['blue','red'],label='comparación por genero')
plt.title('Calificación promedio por Genero')
plt.xlabel('Genero')
plt.ylabel('Calificación Promedio')
plt.legend()
plt.show()
```
