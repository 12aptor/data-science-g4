```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('titanic')
df
```

```python
df.dtypes
```

```python
df.isnull().sum()
```

```python
sns.heatmap(
    data= df.corr(numeric_only=True),
    cmap=sns.diverging_palette(20,230,as_cmap=True),
    center = 0,
    vmin = -1,
    vmax = 1,
    linewidths=0.5,
    cbar_kws={'shrink':0.5},
    annot=True
)
```

# IMPUTAMOS DATOS FALTANTES

```python
df_cleaned = df.copy()
```

```python
df_cleaned['age'] = df_cleaned['age'].fillna(round(df_cleaned['age'].mean()))
```

```python
df_cleaned.isnull().sum()
```

```python
sns.heatmap(
    data= df_cleaned.corr(numeric_only=True),
    cmap=sns.diverging_palette(20,230,as_cmap=True),
    center = 0,
    vmin = -1,
    vmax = 1,
    linewidths=0.5,
    cbar_kws={'shrink':0.5},
    annot=True
)
```

# COMPARAMOS CORRELACIONES CON NULOS Y SIN NULOS EN AGE

```python
corr_matrix = df.corr(numeric_only=True)
corr_age = corr_matrix[['age']].sort_values(by='age',ascending=False)
sns.heatmap(
    data=corr_age,
    cmap=sns.diverging_palette(20, 230, as_cmap=True),
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={'shrink': 0.5},
    annot=True
)
plt.title("Correlación de 'age' con otras variables")
plt.show()
```

```python
corr_matrix = df_cleaned.corr(numeric_only=True)
corr_age = corr_matrix[['age']].sort_values(by='age',ascending=False)
sns.heatmap(
    data=corr_age,
    cmap=sns.diverging_palette(20, 230, as_cmap=True),
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={'shrink': 0.5},
    annot=True
)
plt.title("Correlación de 'age' con otras variables")
plt.show()
```
