# IMPORTAMOS EL MODELO Y EL TOKENIZER DE MISTRAL

```python
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline

model_id = "mistralai/Mistral-7B-Instruct-v0.1"  # Puedes cambiarlo por otro si necesitas

# Cargamos el tokenizer y el modelo
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype="auto", device_map="auto")
```

# CREAMOS EL PIPELINE PARA UN CHATBOT

```python
chatbot = pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    max_new_tokens=100,
    temperature=0.7,
    top_p=0.9,
    do_sample=True,
    device_map="auto"
)
```

# PROBAMOS EL MODELO

```python
print("🤖 Chatbot LLM (escribe 'salir' para terminar)")

history = ""

while True:
    user_input = input("Tú: ")
    if user_input.lower() == "salir":
        break

    prompt = f"<s>[INST] {user_input} [/INST]"  # Formato Instruct
    result = chatbot(prompt)[0]['generated_text']
    response = result.split("[/INST]")[-1].strip()

    print("Bot:", response)
```
