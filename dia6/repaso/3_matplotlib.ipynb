```python
import matplotlib.pyplot as plt
```

```python
import numpy as np
x = np.linspace(0,5,11)
x
```

```python

y = x**2
y
```

```python
plt.plot(x,y,color='red',linestyle='--',marker='o',label='x**2')
plt.title('Relacion entre X y Y')
plt.xlabel('Variable x')
plt.ylabel('Variable y')
plt.legend()
plt.show()
```

```python
plt.barh(categorias,valores,color='green')
plt.show()
```

# SUBPLOTS

```python

x = np.linspace(0,5,11)
y = x**2

plt.subplot(1,2,1)
plt.plot(x,y,'b--')

plt.subplot(1,2,2)
plt.hist(y)

plt.show()
```

```python
fig,axes = plt.subplots(2,4)
axes[0,0].plot(x,np.cos(x))
axes[0,1].plot(x,np.sin(x),'yellow')
axes[0,2].plot(x,np.tan(x),'r')
axes[0,3].plot(x,np.cos(x)**2,'g')
axes[1,0].plot(x,np.cos(y),'b')
axes[1,1].plot(x,np.sin(y),'purple')
axes[1,2].plot(x,np.tan(y),'orange')
axes[1,3].plot(x,np.cos(y)**2)

fig.tight_layout()
plt.show()
```

```python
import pandas as pd
df_books = pd.read_csv('https://github.com/cesarmayta/datasets/raw/refs/heads/main/libros.csv')
df_books.head(2)
```

```python
 precio_promedio = df_books.groupby('Year')['Price'].mean().reset_index()
 x = precio_promedio['Year']
 y = precio_promedio['Price']
```

```python
plt.plot(x,y,color='green',label='Evolucion Precio')
plt.title('Precio promedio por año')
plt.xlabel('Año')
plt.ylabel('Precio Promedio')
plt.legend()
plt.show()
```
