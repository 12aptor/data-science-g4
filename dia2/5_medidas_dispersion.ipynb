```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}

penguins_df = sns.load_dataset('penguins')
penguins_df.dropna(inplace=True)
penguins_df
```

# MEDIDAS DE DISPERSIÓN

# RANGO

```python
# maximo
penguins_df.max(numeric_only=True)
```

```python
# minimos
penguins_df.min(numeric_only=True)
```

```python
rango = penguins_df.max(numeric_only=True) - penguins_df.min(numeric_only=True)
rango
```

# DESVIACIÓN STANDARD

```python
penguins_df.std(numeric_only=True)
```

```python
penguins_df.mean(numeric_only=True) + penguins_df.std(numeric_only=True)
```

# RANGO INTERCUARTILICO

## Q1 - 25%

```python
penguins_df.quantile(0.25,numeric_only=True)
```

# Q3 - 75%

```python
penguins_df.quantile(0.75,numeric_only=True)
```

```python
rango_intercuartilico = penguins_df.quantile(0.75,numeric_only=True) - penguins_df.quantile(0.25,numeric_only=True)
rango_intercuartilico
```

```python
(
    penguins_df
    .quantile(q=[0.75,0.50,0.25],numeric_only=True)
    .transpose()
    .rename_axis('metric')
    .reset_index()
    .assign(
        iqr = lambda df: df[0.75] - df[0.25]
    )
)
```

# GRAFICAMOS TODAS LAS MEDIDAS

```python
variable = 'bill_depth_mm'

sns.histplot(
    data=penguins_df,
    x=variable
)

# MEDIA
plt.axvline(
    x=penguins_df[variable].mean(),
    color='red',
    linestyle='dashed',
    linewidth=2
)

# MEDIANA
plt.axvline(
    x=penguins_df[variable].median(),
    color='yellow',
    linestyle='dashed',
    linewidth=2
)

# MODA
plt.axvline(
    x=penguins_df[variable].mode().values[0],
    color='green',
    linestyle='dashed',
    linewidth=2
)

# Q 25
plt.axvline(
    x=penguins_df[variable].quantile(0.25),
    color='black',
    linestyle='dashed',
    linewidth=2
)

# Q 75
plt.axvline(
    x=penguins_df[variable].quantile(0.75),
    color='black',
    linestyle='dashed',
    linewidth=2
)
```
