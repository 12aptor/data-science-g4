```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# GENERAMOS DATOS  PARA NUESTRO MODELO

```python
# Datos de entrenamiento (X: características, y: etiquetas)
np.random.seed(42)
X = np.random.randn(10, 2)  # 500 muestras con 2 características
y = (X[:, 0] * X[:, 1] > 0).astype(np.float32)  # Clase 1 si x1*x2 > 0, sino Clase 0
```

```python
class BackPropagation:
    def __init__(self, learning_rate=0.1):
        # Inicializamos pesos aleatorios
        self.w1, self.w2 = np.random.rand(), np.random.rand()  # Pesos de entrada
        self.b = np.random.rand()  # Sesgo
        self.lr = learning_rate  # Tasa de aprendizaje

    def sigmoid(self, x):
        return 1 / (1 + np.exp(-x))  # Función sigmoide

    def sigmoid_derivative(self, x):
        return x * (1 - x)  # Derivada de la sigmoide

    def forward(self, x1, x2):
        # Propagación hacia adelante
        self.z = x1 * self.w1 + x2 * self.w2 + self.b  # Suma ponderada
        self.y_pred = self.sigmoid(self.z)  # Aplicamos sigmoide
        return self.y_pred

    def backward(self, x1, x2, y_true):
        # Cálculo del error
        error = y_true - self.y_pred

        # Derivadas parciales
        d_loss_d_pred = -2 * error  # Derivada de la pérdida con respecto a la predicción
        d_pred_d_z = self.sigmoid_derivative(self.y_pred)  # Derivada de la sigmoide

        # Gradientes de los pesos
        d_loss_d_w1 = d_loss_d_pred * d_pred_d_z * x1
        d_loss_d_w2 = d_loss_d_pred * d_pred_d_z * x2
        d_loss_d_b = d_loss_d_pred * d_pred_d_z * 1  # Sesgo

        # Actualización de pesos
        self.w1 -= self.lr * d_loss_d_w1
        self.w2 -= self.lr * d_loss_d_w2
        self.b -= self.lr * d_loss_d_b

        # Mostrar mensajes
        print(f"Error: {error:.4f}, w1: {self.w1:.4f}, w2: {self.w2:.4f}, b: {self.b:.4f}")

    def train(self, X, y, epochs=10):
        for epoch in range(epochs):
            print(f"\n🔹 Época {epoch+1}")
            for i in range(len(X)):
                x1, x2 = X[i]
                y_true = y[i]

                # Forward y Backpropagation
                self.forward(x1, x2)
                self.backward(x1, x2, y_true)
```

```python
# Entrenamos la red neuronal
nn = BackPropagation(learning_rate=0.5)
nn.train(X, y, epochs=5)
```

# BACKPROPAGATION USANDO TENSORFLOW CON ANN Redes Neuronales Artificales

```python
import tensorflow as tf
model = tf.keras.Sequential([
    tf.keras.layers.Dense(10,activation='relu'),
    tf.keras.layers.Dense(1,activation='sigmoid')
])

model.compile(optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
              loss='binary_crossentropy',
              metrics=['accuracy'])
history = model.fit(X,y,epochs=5,verbose=1)
```
