```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}

penguins_df = sns.load_dataset('penguins')
penguins_df.dropna(inplace=True)
penguins_df
```

# RELACIONES ENTRE 2 VARIABLES NÚMERICAS

# SCATTERPLOT

```python
sns.scatterplot(
    data=penguins_df,
    x='bill_length_mm',
    y='bill_depth_mm',
    alpha=1/2,
    s=100
)
```

```python
sns.scatterplot(
    data=penguins_df,
    x='flipper_length_mm',
    y='body_mass_g',
    alpha=1/2,
    s=100
)
```

```python
sns.scatterplot(
    data=penguins_df,
    x='bill_depth_mm',
    y='body_mass_g',
    hue='species',
    alpha=1/2,
    s=100
)
```

## DISPLOT

```python
sns.displot(
    data=penguins_df,
    x='bill_length_mm',
    y='body_mass_g',
    rug=True
)
```

```python

sns.displot(
    data=penguins_df,
    x='flipper_length_mm',
    y='body_mass_g',
    rug=True
)
```

```python

sns.displot(
    data=penguins_df,
    x='flipper_length_mm',
    y='body_mass_g',
    rug=True,
    kind='kde'
)
```
