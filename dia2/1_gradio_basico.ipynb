# CREAMOS INTERFAZ GRAFICA WEB CON GRADIO

```python
import gradio as gr


#creamos una función
def saluda(nombre):
  return 'Hola ' + nombre

demo = gr.Interface(fn=saluda,inputs='text',outputs='text')

demo.launch()
```

# APP PARA SUMA DE 2 NÚMEROS

```python
import gradio as gr

def suma(n1,n2):
  resultado = int(n1) + int(n2)
  return resultado

demo = gr.Interface(
    fn=suma,
    inputs=[
        gr.Textbox(label='Número 1',placeholder='0'),
        gr.Textbox(label='Número 2',placeholder='0')
    ],
    outputs=gr.Textbox(label='Resultado')
)

demo.launch()
```
