```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
sns.set_style(style='whitegrid')
sns.set_context(context='notebook')
plt.rcParams['figure.figsize'] = (11, 9.4)

penguin_color = {
    'Adelie': '#ff6602ff',
    'Gentoo': '#0f7175ff',
    'Chinstrap': '#c65dc9ff'
}

penguins_df = sns.load_dataset('penguins')
penguins_df.dropna(inplace=True)
penguins_df
```

# MEDIA

```python
# usando numpy
np.mean(penguins_df.bill_depth_mm)
```

```python
# usando pandas
penguins_df.bill_depth_mm.mean()
```

```python
# media de todos las variables númericas
penguins_df.mean(numeric_only=True)
```

# MEDIANA

```python
penguins_df.median(numeric_only=True)
```

# MODA

```python
penguins_df['bill_depth_mm'].mode().values[0]
```

```python
penguins_df.mode(numeric_only=True)
```

# GRAFICAMOS MEDIDAS DE TENDENCIA CENTRAL DE BILL_DEPTH_MM

```python
variable = 'bill_depth_mm'

sns.histplot(
    data=penguins_df,
    x=variable
)

# MEDIA
plt.axvline(
    x=penguins_df[variable].mean(),
    color='red',
    linestyle='dashed',
    linewidth=2
)

# MEDIANA
plt.axvline(
    x=penguins_df[variable].median(),
    color='yellow',
    linestyle='dashed',
    linewidth=2
)

# MODA
plt.axvline(
    x=penguins_df[variable].mode().values[0],
    color='green',
    linestyle='dashed',
    linewidth=2
)
```
