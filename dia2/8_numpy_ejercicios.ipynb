# Paso 1: Genera un array con 30 valores aleatorios entre 15 y 40 (temperaturas diarias en °C)

```python
temperaturas = np.random.uniform(15,40,30)
temperaturas
```

# PASO 2- DIVIDIMOS LA DATA EN UNA MATRIZ DE 5X6 PARA TENER LA TEMPERATURA POR SEMANA

```python
matriz_temp = temperaturas.reshape(5,6)
matriz_temp
```
