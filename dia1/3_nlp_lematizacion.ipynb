## **Lematización:** Proceso para encontrar la raíz linguística de una palabra

# CASO 1 : USANDO NLTK - INGLES

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
lemmatizer = WordNetLemmatizer()

palabras = ['running','flies','better']

for palabra in palabras:
  lema = lemmatizer.lemmatize(palabra,wordnet.VERB)
  print(f'palabra: {palabra} -> Lema: {lema}')
```

```python
texto = "hoy trabajaré hasta tarde"
```

```python
import spacy
nlp = spacy.load('es_core_news_sm')

doc = nlp(texto)

#obtener lematización
texto_lematizado =  " ".join([token.lemma_ for token in doc])
print(texto_lematizado)
```
