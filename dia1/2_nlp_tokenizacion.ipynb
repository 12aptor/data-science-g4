# PROCESO DE NORMALIZACION DE TEXTOS
## 1 - TOKENIZACIÓN
La **tokenización** es uno de los primeros pasos fundamentales en el procesamiento del lenguaje natural (**NLP**, por sus siglas en inglés). Consiste en dividir un texto en unidades más pequeñas llamadas **tokens**. Estos tokens suelen ser palabras, caracteres o sub-palabras.

```python
print(r'esto es una linea \n esto es otra')
```

```python
texto = """El Procesamiento de Lenguaje Natural (NLP) es una rama de la inteligencia artificial (IA) que se enfoca en la interacción entre los computadores y el lenguaje humano (natural).
           NLP permite a las máquinas (computadoras y programas) entender, interpretar y generar lenguaje de manera similar a como lo hacen los humanos."""
print(texto)
```

# CASO 1: TOKENIZACIÓN CON PYTHON PURO USANDO SPLIT

```python
import re
corpus = re.split(f' ',texto)
print(corpus)
```

# CASO 2: TOKENIZACIÓN USANDO EXPRESIONES REGULARES

```python
corpus = re.split(r'[ \W\t\n]+',texto)
print(corpus)
```

# CASO 3: TOKENIZACIÓN USANDO NLTK

```python
#usamos expresiones regulares para tokenizar
texto = 'En los E.U. esa postal vale $15.50 ...'
print(re.split(r'[ \W\t\n]+', texto))
```

```python
#usamos nltk para tokenizar
import nltk
pattern = r'''(?x)                 # Permite comentarios en la expresión regular
              (?:[A-Z]\.)+          # Abreviaturas, e.g., U.S.A.
              | \w+(?:-\w+)*        # Palabras con guiones opcionales
              | \$?\d+(?:\.\d+)?%?  # Monedas y porcentajes, e.g., $12.40, 82%
              | \.\.\.              # Elipsis
              | [.,;"'?():-_`]      # Caracteres de puntuación como tokens separados
           '''
corpus = nltk.regexp_tokenize(texto,pattern)
print(corpus)
```
