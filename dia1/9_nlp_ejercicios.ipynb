# EJERCICIO 1- CLASIFICADOR DE EMAIL SPAM

## PASO 1 - IMPORTAMOS LIBRERIAS

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab') # Added download for punkt_tab
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.classify import NaiveBayesClassifier
from nltk.corpus import stopwords
```

## PASO 2 - CREAMOS UN DATASET DE SPAM

```python
dataset = [
    ("Congratulations! You've won a free ticket!", "spam"),
    ("Call now to claim your prize", "spam"),
    ("This is not spam, please read the document", "ham"),
    ("Let's meet for lunch tomorrow", "ham"),
    ("Free entry in contest!", "spam"),
    ("See you at the meeting", "ham")
]
```

## PASO 3 - PREPROCESAMIENTO Y EXTRACCION DE CARACTERISTICAS

```python
def limpiar_texto(texto):
    stop_words = set(stopwords.words('english'))
    palabras = word_tokenize(texto.lower())
    palabras = [p for p in palabras if p.isalpha() and p not in stop_words]
    return dict([(palabra, True) for palabra in palabras])
```

```python
nltk.download('stopwords')
```

```python
datos_entrenamiento = [(limpiar_texto(texto), etiqueta) for texto, etiqueta in dataset]
```

## PASO 4 - ENTRENAMOS EL MODELO CON NLTK

```python
modelo = NaiveBayesClassifier.train(datos_entrenamiento)
```

## PASO 5 - PROBAMOS EL CLASIFICADOR

```python
texto_prueba = "Congratulations! Free prize waiting for you"
resultado = modelo.classify(limpiar_texto(texto_prueba))
print(f"Texto: {texto_prueba}")
print(f"Clasificación: {resultado}")
```

# EJERCICIO 2 - ANALISIS DE SENTIMIENTOS

## PASO 1 IMPORTAMOS LIBRERIAS

```python
import nltk
nltk.download('punkt')
nltk.download('stopwords')

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy
```

## PASO 2 - CREAMOS EL DATASET

```python
dataset = [
    ("I love this product, it's amazing!", "pos"),
    ("This movie was fantastic and enjoyable", "pos"),
    ("What a great experience!", "pos"),
    ("I hate this service, very disappointing", "neg"),
    ("This is the worst thing I have bought", "neg"),
    ("Completely useless and bad", "neg")
]
```

## PASO 3 - PROCESAMOS LA DATA

```python
def extraer_caracteristicas(texto):
    stop_words = set(stopwords.words('english'))
    palabras = word_tokenize(texto.lower())
    palabras_limpias = [p for p in palabras if p.isalpha() and p not in stop_words]
    return dict([(palabra, True) for palabra in palabras_limpias])
```

```python
datos_entrenamiento = [(extraer_caracteristicas(texto), etiqueta) for texto, etiqueta in dataset]
```

## PASO 4 - ENTRENAMIENTO DEL MODELO

```python
modelo = NaiveBayesClassifier.train(datos_entrenamiento)
```

## PASO 5- PROBAMOS EL MODELO

```python
texto_prueba = "I really love this amazing movie!"
resultado = modelo.classify(extraer_caracteristicas(texto_prueba))
print(f"Texto: {texto_prueba}")
print(f"Sentimiento detectado: {resultado}")
```
