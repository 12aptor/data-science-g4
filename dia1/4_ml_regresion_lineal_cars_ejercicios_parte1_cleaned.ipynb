# CONVERTIR DATOS CATEGORICOS A NÚMERICOS

```python
# Converting column data types
df['model_year'] = df['model_year'].astype('int64')

# Remove non-numeric characters from 'milage' and convert to float
df['milage'] = df['milage'].str.replace(',', '').str.replace(' mi.', '', regex=False).astype('float64')

# Remove non-numeric characters from 'price' and convert to float
df['price'] = df['price'].str.replace('$', '', regex=False).str.replace(',', '', regex=False).astype('float64')

# Print the data types to verify the changes
df.dtypes
```

# REVISAMOS COLUMNAS CATEGORICAS

```python
categorical_columns = df.select_dtypes(object).columns
categorical_columns
```

```python
for cc in categorical_columns:
  print("*"*50)
  print(df[cc].value_counts())
```

```python
# prompt: la columna accident es categorica y tiene estos valores 'None reported' y 'At least 1 accident or damage reported' codificala 0 para el primero y 1 para el 2do caso y conviertela a int64

df['accident'] = df['accident'].map({'None reported': 0, 'At least 1 accident or damage reported': 1})
df['accident'] = df['accident'].astype('int64')
```

```python
# prompt: analiza la columna transmission y encuentra un patron que permita codificar solo el valores 0 y 1 por el tipo de transimision comun po rjemplo une todoso lso que digan manual en 0 y automatico en 1

import numpy as np
# Analyze the 'transmission' column and create a binary encoding
# Identify values that represent 'Manual' and 'Automatic' transmissions
manual_keywords = ['Manual']
automatic_keywords = ['Automatic']

# Create the new binary column
df['transmission_encoded'] = np.nan  # Initialize with NaN

# Encode Manual transmissions as 0
for keyword in manual_keywords:
  df.loc[df['transmission'].str.contains(keyword, case=False, na=False), 'transmission_encoded'] = 0

# Encode Automatic transmissions as 1
for keyword in automatic_keywords:
    df.loc[df['transmission'].str.contains(keyword, case=False, na=False), 'transmission_encoded'] = 1

# Verify the changes
print("\nValue counts for original 'transmission' column:")
print(df['transmission'].value_counts())
print("\nValue counts for new 'transmission_encoded' column:")
print(df['transmission_encoded'].value_counts(dropna=False)) # Include NaN counts
print("\nFirst few rows with the new column:")
print(df[['transmission', 'transmission_encoded']].head())

# You might want to handle any remaining NaN values in 'transmission_encoded'
# depending on your requirements (e.g., drop rows or fill with a placeholder)
# Example: Drop rows where transmission_encoded is NaN
# df.dropna(subset=['transmission_encoded'], inplace=True)
```

```python
# prompt: renombra la columna transmission_encoded por automatic y reemplaza los Nan por 0

df = df.rename(columns={'transmission_encoded': 'automatic'})
df['automatic'] = df['automatic'].fillna(0)
print("\nValue counts for the renamed and filled 'automatic' column:")
print(df['automatic'].value_counts(dropna=False))
print("\nFirst few rows with the renamed column and filled NaNs:")
print(df[['transmission', 'automatic']].head())
```

```python
# prompt: analiza la columna engine y encuentra el nro que esta antes de la letra "L" por ejemplo "3.7L" y solo captura el numero y conviertelo a float64

df['engine'] = df['engine'].str.extract(r'(\d+\.?\d*)L').astype('float64')
```
