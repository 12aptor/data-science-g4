# MODELO DE REGRESION LINEAL SIMPLE

## ¿Qué es la regresión lineal?

La **regresión lineal** es una técnica estadística y de aprendizaje supervisado que modela la relación entre una variable dependiente (también llamada variable de respuesta o `Y`) y una o más variables independientes (predictoras o `X`).

El objetivo principal es **predecir** el valor de `Y` a partir de los valores de `X`, asumiendo que existe una relación **lineal** entre ellas.

---

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/housing/housing.data',header=None,sep='\s+')
df.columns =  ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT', 'MEDV']
df.head(5)
```

# CORRELACIONES

```python
cols = ['DIS','INDUS','CRIM', 'RM', 'MEDV']
sns.heatmap(
    data=df[cols].corr(),
    cmap=sns.diverging_palette(20, 230, as_cmap=True),
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={"shrink": 0.5},
    annot=True
)
```

```python
sns.scatterplot(data=df,x=df['RM'],y=df['MEDV'])
```

## 🧮 1. Regresión Lineal Simple

Es el caso más básico donde se relaciona **una sola variable independiente** `X` con la variable dependiente `Y`.

### 📐 Ecuación:

$$
Y = \beta_0 + \beta_1 X + \varepsilon
$$

- \( Y \): variable dependiente  
- \( X \): variable independiente  
- \( \beta_0 \): intercepto o término independiente  
- \( \beta_1 \): pendiente o coeficiente de regresión  
- \( \varepsilon \): término de error (ruido)

---

# CREANDO DEL MODELO DE REGRESIÓN LIENAL CON SKLEARN

## PASO 1 - DEFINIMOS VARIABLE X Y Y

X = RM(PROMEDIO DE HABITACIONES EN EL DEPARTAMENTO)
Y = MEDV(VALOR PROMEDIO DEL DEPARTAMENTO)

```python
X = df['RM'].values.reshape(-1,1)
y = df['MEDV'].values.reshape(-1,1)
```

## PASO 2 IMPORTAMOS LIBRERIAS

```python
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
```

## PASO 3 ESTANDARIZAMOS LOS VALORES

```python
sc_x = StandardScaler()
sc_y = StandardScaler()

X_std = sc_x.fit_transform(X)
y_std = sc_y.fit_transform(y)
```

## PASO 4 - CREAMOS EL MODELO

```python
slr = LinearRegression()
slr.fit(X_std,y_std)
slr.predict(X_std)
```

# GRAFICAMOS LOS RESULTADOS

```python
plt.scatter(X_std,y_std)
plt.plot(X_std,slr.predict(X_std),color='red')
plt.ylabel('MEDV')
plt.xlabel('RM')
```

# PROBAMOS PREDICCIONES

```python
num_hab = 7.185
num_hab_std = sc_x.transform(np.array([num_hab]).reshape(-1,1))
print(f" EL PRECIO PARA UN DPTO CON {num_hab}  ES DE : {sc_y.inverse_transform(slr.predict(num_hab_std))}")
```

# ENTRENAMOS NUESTRO MODELO DIVIENDO EL DATASET EN TRAIN Y TEST

```python
from sklearn.model_selection import train_test_split

data = df.copy()
data.head(2)
```

```python
X = data['RM'].values.reshape(-1,1)
y = data['MEDV'].values.reshape(-1,1)
```

# DIVIDIMOS EL DATASET EN 70% PARA ENTRENAMIENTO Y 30% PARA PRUEBAS

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.30)
```

```python
sc_x = StandardScaler()
X_train = sc_x.fit_transform(X_train)
X_test = sc_x.fit_transform(X_test)

sc_y = StandardScaler()
y_train = sc_y.fit_transform(y_train)
y_test = sc_y.fit_transform(y_test)
```

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
y_pred.shape
```

# EVALUAMOS EL MODELO CON METRICAS MSE Y R2

MSE nos dice cuánto se equivoca el modelo.(0 es lo ideal)

R² nos dice qué tanto del comportamiento del dato ha logrado explicar el modelo.(1 es lo ideal)

```python
import sklearn.metrics as metrics
```

```python
#error cuadratico medio
mse = metrics.mean_squared_error(y_test,y_pred)
#R2 Coeficiente de determinación
r2 = metrics.r2_score(y_test,y_pred)

print(f'El error cuadratico medio(MSE) es de {round(mse,4)}')
print(f'El coeficiente de determinación(R2) es de {round(r2,4)}')
```

## 🧮 2. Regresión Lineal Múltiple

Extiende la regresión simple para incluir **dos o más variables independientes**:

$$
Y = \beta_0 + \beta_1 X_1 + \beta_2 X_2 + \cdots + \beta_n X_n + \varepsilon
$$

```python
cols = ['DIS','INDUS','CRIM', 'RM', 'MEDV']
sns.heatmap(
    data=df[cols].corr(),
    cmap=sns.diverging_palette(20, 230, as_cmap=True),
    center=0,
    vmin=-1,
    vmax=1,
    linewidths=0.5,
    cbar_kws={"shrink": 0.5},
    annot=True
)
```

```python
sns.scatterplot(data=df,x=df['DIS'],y=df['MEDV'])
```

# CREAMOS MODELO DE REGRESIÓN LINEAL MULTIPLE

```python
X = df[['RM','DIS','INDUS','CRIM']].values
y = df['MEDV'].values.reshape(-1,1)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20)
```

```python
sc_x = StandardScaler()
X_train = sc_x.fit_transform(X_train)
X_test = sc_x.fit_transform(X_test)

sc_y = StandardScaler()
y_train = sc_y.fit_transform(y_train)
y_test = sc_y.fit_transform(y_test)

model = LinearRegression()
model.fit(X_train,y_train)

y_pred = model.predict(X_test)
y_pred.shape
```

# EVALUAMOS EL MODELO

```python
import sklearn.metrics as metrics

#erro cuadratico medio
mse = metrics.mean_squared_error(y_test,y_pred)
#R2 Coeficiente de determinación
r2 = metrics.r2_score(y_test,y_pred)

print(f'El error cuadratico medio(MSE) es de {round(mse,4)}')
print(f'El coeficiente de determinación(R2) es de {round(r2,4)}')
```

# GRAFICAMOS LO REAL VS LOS PREDICHO

```python
# Invertimos la escala para llevar y_test y y_pred a sus valores originales
y_test_real = sc_y.inverse_transform(y_test)
y_pred_real = sc_y.inverse_transform(y_pred)

# Gráfico
plt.figure(figsize=(8, 6))
sns.scatterplot(x=y_test_real.flatten(), y=y_pred_real.flatten(), color='purple', alpha=0.6)
plt.plot([y_test_real.min(), y_test_real.max()],
         [y_test_real.min(), y_test_real.max()],
         color='red', linestyle='--', label='Línea ideal')

plt.xlabel('Valores Reales de MEDV')
plt.ylabel('Valores Predichos de MEDV')
plt.title('Comparación: Valores Reales vs Predichos')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
