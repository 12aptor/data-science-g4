# IMPORTAMOS Y DESCARGAMOS UN TEXTO EN ESPAÑOL

```python
import nltk
nltk.download('cess_esp')
```

# CREAMOS UN CORPUS(COJUNTO DE TEXTOS)

```python
corpus = nltk.corpus.cess_esp.sents()
corpus[1]
```

# REVISAMOS EL TOTAL DE TEXTOS DEL CORPUS

```python
len(corpus)
```

## CONVERTIMOS EL CORPUS EN UNA SOLA LISTA DE PALABRAS

```python
total_corpus = [w for l in corpus for w in l]
total_corpus[0]
```

```python
len(total_corpus)
```

# FILTRANDO TEXTOS CON EXPRESIONES REGULARES

# BUSCAR TODAS LAS PALABRAS QUE CONTENGAS 'ES'

```python
import re
arr = [w for w in total_corpus if re.search('es',w)]
arr[:10]
```

# BUSCAR TODAS PALABRAS QUE TENGAS J T DENTRO DE LA PALABRA EN ESE ORDEN

```python
arr = [w for w in total_corpus if re.search('^..j..t..$',w)]
arr
```
