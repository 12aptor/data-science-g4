# COMO CREAMOS UNA RED NEURONAL CON PYTORCH

## 1 IMPORTAMOS LIBRERIAS

```python
import torch
import torch.nn as nn
import torch.optim as optim
```

## 2 - CREAMOS UNA CLASE QUE HEREDE DE LA CLASE NN.MODULE

```python
class SimpleNN(nn.Module):
  def __init__(self):
    super(SimpleNN,self).__init__()
    self.fc = nn.Linear(2,1) # creamos una capa con 2 neuronas de entrada y 1 de salida

  #creamos el metodo forward
  def forward(self,x):
    return self.fc(x)
```

## 3 - CREAMOS UN OBJETO DE LA CLASE SIMPLENN Y DEFINIMOS FUNCION DE PERDIDA Y OPTIMIZADOR

```python
model = SimpleNN()
criterion = nn.MSELoss() # error cuadratico medio
optimizer = optim.SGD(model.parameters(),lr=0.01) # descenso del gradiente estocástico(SGD)
```

## 4 CREAMOS UN DATASET DE PRUEBA PARA ENTRENAR NUESTRO MODELO

```python
inputs = torch.tensor([[1.0,2.0],[2.0,3.0]],dtype=torch.float32)
inputs
```

```python
labels = torch.tensor([[1.0],[2.0]],dtype=torch.float32)
labels
```

## 5 - ENTRENAMOS EL MODELO

```python
epochs = 10
for epoch in range(epochs):
  outputs = model(inputs) # hacemos una predicción con la red neuronal
  loss = criterion(outputs,labels) # calculamos el error o perdida

  #backpropagration y optimización
  optimizer.zero_grad() # reiniciar los gradientes para evitar acumulaciones incorrectas
  loss.backward() # calcular los gradientes
  optimizer.step() # actualizar los pesos del modelo
  print(f"Epoch {epoch+1}/{epochs}, Loss: {loss.item():.4f}")

  if(epoch+1) % 10 == 0:
    print(f'resultado final : Epoch {epoch+1}, Loss: {loss.item():.4f}')
```

## 6 PREDICCIONES DEL MODELO

```python
import numpy as np
print(f'valores predichos : {np.round(model(inputs).detach().numpy())}')
print(f'valores reales :{labels}')
```
