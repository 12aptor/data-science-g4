# CREAMOS NUESTRA RED NEURONAL

```python
class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, stride=2, padding=1)
        self.pool = nn.MaxPool2d(kernel_size=2)
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=32, kernel_size=3, padding=1)
        self.flatten = nn.Flatten()

        # Calcular el tamaño correcto automáticamente
        with torch.no_grad():
            sample_input = torch.zeros(1, 3, 64, 64)  # Simulación de entrada
            sample_output = self._get_flattened_size(sample_input)

        self.fc1 = nn.Linear(sample_output, 128)  # Usar el tamaño calculado
        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(128, 1)  # Dos clases

    def _get_flattened_size(self, x):
        x = self.pool(nn.ReLU()(self.conv1(x)))
        x = self.pool(nn.ReLU()(self.conv2(x)))
        return x.numel()  # Obtener el número total de elementos después del flatten

    def forward(self, x):
        x = self.pool(nn.ReLU()(self.conv1(x)))
        x = self.pool(nn.ReLU()(self.conv2(x)))
        x = self.flatten(x)
        print(f"Shape después de flatten: {x.shape}")  # Depuración
        x = nn.ReLU()(self.fc1(x))
        x = self.dropout(x)
        x = self.fc2(x)
        return x
```

```python
import torch
import torch.nn as nn

class CNN(nn.Module):
    def __init__(self):
        super(CNN, self).__init__()
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=(3, 3), stride=(2, 2), padding=1)
        self.pool = nn.MaxPool2d(kernel_size=(2, 2))
        self.conv2 = nn.Conv2d(in_channels=32, out_channels=32, kernel_size=(3, 3), padding=1)
        self.flatten = nn.Flatten()

        # Esta dimensión cambiará una vez calcules la salida correcta
        self.fc1 = nn.Linear(32 * 16 * 16, 128)  # Temporalmente, 32 * 16 * 16 es un ejemplo

        self.dropout = nn.Dropout(0.5)
        self.fc2 = nn.Linear(128, 1)

    def forward(self, x):
        x = self.pool(nn.ReLU()(self.conv1(x)))
        x = self.pool(nn.ReLU()(self.conv2(x)))

        print(x.shape)  # Imprimir la forma de la salida aquí para ver cuántas características hay

        x = self.flatten(x)
        x = nn.ReLU()(self.fc1(x))
        x = self.dropout(x)
        x = torch.sigmoid(self.fc2(x))
        return x
```

## CONFIGURAMOS EL MODELO

```python
criterion = nn.BCEWithLogitsLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)
```

## ENTRENAMOS NUESTRO MODELO

```python
def train(model, train_loader, criterion, optimizer, epochs=10):
    model.train()
    for epoch in range(epochs):
        running_loss = 0.0
        for inputs, labels in train_loader:
            inputs, labels = inputs.to(device), labels.to(device)
            #labels = labels.long().squeeze()
            labels = labels.float().squeeze()

            optimizer.zero_grad()

            outputs = model(inputs)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            running_loss += loss.item()
             # Calcular el accuracy
            labels = labels.long().squeeze()  # Asegurar etiquetas correctas
            _, predicted = torch.max(outputs, 1)  # Obtiene las clases predichas
            total += labels.size(0)
            correct += (predicted == labels).sum().item()  # Comparar correctamente


        epoch_loss = running_loss / len(train_loader)
        epoch_accuracy = 100 * correct / total
        print(f"Epoch {epoch+1}, Loss: {epoch_loss:.4f}, Accuracy: {epoch_accuracy:.2f}%")
```

```python
train(model, train_loader, criterion, optimizer, epochs=2)
```
