# ANALISIS DE DATOS FALTANTES

## TOTAL DE FALTANTES

```python
restantes = df.shape[0] -  df.missing.number_missing()
restantes
```

## ANALISIS DE VARIABLES FALTANTES

```python
df.missing.missing_variable_plot()
```

## GRAFICAMOS PROPORCIÓN DE DATOS FALTANTES

```python
import missingno
df.missing.sort_variables_by_missingness().pipe(missingno.matrix)
```

# IMPUTACIÓN POR MEDIDAS DE TENDENCIA CENTRAL

## IMPUTACIÓN DE LA EDAD

```python
round(df.age.mean())
```

```python
df_age_mean = df.copy()
df_age_mean['age'] = df_age_mean['age'].fillna(round(df_age_mean['age'].mean()))
```

```python
df_age_mean.missing.missing_variable_plot()
```

## VALIDAMOS LA IMPUTACIÓN

```python
print(f" media original : {round(df['age'].mean())} | media sin nulos : {round(df_age_mean['age'].mean())}")
```
