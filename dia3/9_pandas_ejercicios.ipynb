```python
import pandas as pd
df_books = pd.read_csv('https://github.com/cesarmayta/datasets/raw/refs/heads/main/libros.csv')
df_books.shape
```

### FILTRAR POR GENERO 'FICTION' Y RATING > 4
### MOSTRAR SOLO NAME,AUTHOR, USER RATING, PRICE

```python
df_books[(df_books['Genre']=='Fiction') & (df_books['User Rating'] > 4)][['Name','Author','User Rating','Price']]
```

### AGRUPA POR AUTHOR , MUESTRA PRECIO PROMEDIO, TOTAL DE LIBROS

```python
df_books.groupby('Author').agg(
    precio_promedio=('Price','mean'),
    total_libros=('Name','count')
).reset_index()
```
