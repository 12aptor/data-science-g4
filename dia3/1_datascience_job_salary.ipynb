```python
def classify_zona(row):
    if row['region'] != 'Americas':
        return row['region']
    else:
        if row['sub-region'] == 'Latin America and the Caribbean':
            return 'Latin America'
        elif row['sub-region'] == 'Northern America':
            return 'NorthAmerica'
        else:
            return row['region']

df_paises['zona'] = df_paises.apply(classify_zona, axis=1)
display(df_paises.head())
```

```python
df_paises['zona'].value_counts()
```

```python
df_cleaned = df_cleaned.merge(df_paises[['alpha-2', 'zona']], left_on='employee_residence', right_on='alpha-2', how='left')
df_cleaned = df_cleaned.drop(columns='alpha-2')
df_cleaned = df_cleaned.rename(columns={'zona': 'pais_empleado'})
display(df_cleaned.head())
```

```python
data_df_clean = pd.read_csv('job_salary_clean.csv')
data_df_clean
```

```python
categorical_columns = data_df_clean.select_dtypes(object).columns
categorical_columns
```

```python
for cc in categorical_columns:
  print("*"*50)
  print(data_df_clean[cc].value_counts())
```

```python
ordinal_cols = ['company_size']
onehot_cols = list(set(categorical_columns) - set(ordinal_cols))
onehot_cols
```

```python
import sklearn.compose
import sklearn.preprocessing
```

```python
transformer = sklearn.compose.make_column_transformer(
    (sklearn.preprocessing.OrdinalEncoder(),ordinal_cols),
    (sklearn.preprocessing.OneHotEncoder(),onehot_cols),
    remainder='passthrough'
)
transformer
```

```python
data_transformed = transformer.fit_transform(data_df_clean)


```

```python
# si es sparse convertirlo a array
if hasattr(data_transformed, "toarray"):
    data_transformed = data_transformed.toarray()
```

```python
data_transformed_df = pd.DataFrame(
    data_transformed,
    columns=transformer.get_feature_names_out(),
    index=data_df_clean.index
)
```

```python
data_transformed_df
```

```python
data_transformed_df = data_transformed_df.rename(
    columns=lambda x: x.replace("ordinalencoder__", "").replace("onehotencoder__", "").replace("remainder__", "")
)

data_transformed_df
```

```python
data_transformed_df = (
    data_transformed_df.clean_names(
        case_type="snake"
    )
)
data_transformed_df.columns
```

```python
data_transformed_df = data_transformed_df.rename(
    columns=lambda x: x.replace("employment_type_", "tipo_")
    .replace("pais_empleador_", "empleador_")
    .replace("pais_empleado_", "empleado_")
    .replace("experience_level_", "seniority_")
)
data_transformed_df
```

```python
data_transformed_df = data_transformed_df.rename(
    columns={
        "salary_in_usd":"salary"
    }
)
data_transformed_df.columns
```
