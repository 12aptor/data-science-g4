# ARBOLES DE DECISIÓN

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic_df = sns.load_dataset('titanic')
titanic_df.head(10)
```

# EDA

```python
columns_to_drop = ['fare','embarked','class','who','adult_male','deck','embark_town','alive','alone']
titanic_df.drop(columns_to_drop,axis='columns',inplace=True)
titanic_df
```

```python
titanic_df.dropna()
```

# CODIFICACION DE VARIABLES CATEGORICAS

```python
titanic_df = pd.get_dummies(titanic_df,columns=['sex'],drop_first=True)
titanic_df.head()
```

```python
titanic_df['sex_male'] = titanic_df['sex_male'].astype(int)
titanic_df.rename(columns = {'sex_male':'sex'},inplace=True)
titanic_df.head(10)
```

```python
titanic_df.dtypes
```

```python
# ELIMINA LOS VALORES NULOS DEL DATASET Y APLICA ESTE CAMBIO EN EL MISMO DATAFRAME CON EL PARAMETRO INPLACE=TRUE
titanic_df.dropna(inplace=True)
```

```python
titanic_df.isnull().sum()
```

# ENTRENAMIENTO DE MODELO USANDO ARBOLES DE DECISIÓN

# PASO 1 - DEFINIMOS VARIABLES INDEPENDIENTES(X) Y DEPENDIENTES (Y)

```python
X = titanic_df.drop('survived',axis=1)
y = titanic_df['survived']
```

# PASO 2 - CREAMOS EL MODELO DE ARBOL DE DECISIÓN

```python
from sklearn.tree import DecisionTreeClassifier

tree = DecisionTreeClassifier()
```

# PASO 3 - DIVIDIR DATASET EN ENTRENAMIENTO Y PRUEBA

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=00000)
```

# PASO 4 - ENTRENAMOS EL MODELO

```python
tree.fit(X_train,y_train)
```

# EVALUAMOS EL MODELO

```python
from sklearn.metrics import accuracy_score

y_pred = tree.predict(X_test)
accuracy = accuracy_score(y_test,y_pred)
print(f'Precisión del modelo: {accuracy}')
```


# MATRIZ DE CONFUSIÓN

```python
from sklearn.metrics import confusion_matrix

conf_matrix = confusion_matrix(y_test, y_pred)
plt.figure(figsize=(6, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues', xticklabels=['No sobrevivió', 'Sobrevivió'], yticklabels=['No sobrevivió', 'Sobrevivió'])
plt.xlabel('Predicho')
plt.ylabel('Real')
plt.title('Matriz de Confusión')
plt.show()
```
